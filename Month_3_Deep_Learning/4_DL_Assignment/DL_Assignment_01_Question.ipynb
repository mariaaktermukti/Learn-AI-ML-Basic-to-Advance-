# Question 03: [ Marks 15 ]

Why is the perceptron called a linear classifier?
Explain your answer using the concept of linear separability and logical gate examples such as AND, OR, and XOR.

## Write Answer 03:

Perceptron means to a linear classifier because it separates data using a straight line 2D, plane 3D.
It is only limited to classify data that can be linearly separated.

Linear separability means that two classes of data can be separated by a single straight line.
a) If a straight line can divide two classes clearly, the data is linearly separable.

b) If no straight line can separate them, the data is not linearly separable.

A perceptron works only for linearly separable data, so it is called a linear classifier.



AND Gate:

x1:  0 0 1 1

x2:  0 1 0 1

out:0 0 0 1


These points can be separated by a straight line.So, AND gate is linearly separable.




OR gate:

x1 : 0 0  1 1

x2: 0 1 0 1

out: 0 1 1 1

These points can also be separated by a straight line.So, OR gate is linearly separable.




XOR gate:

x1; 0 0 1 1

x2: 0 1 0 1

out: 0 1 1 0

These points cannot be separated by a single straight line.So, XOR gate is not linearly separable.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:
The weight update rule of a perceptron is:

W new= W old+ eta(y-y`)Xi

here,

   W= weight

   eta= learning rate

   y= target output

   y`= predicted output
   



Intuitive: The perceptron learns from its mistakes.The perceptron first makes a prediction. Then it compares the prediction with the correct answer.If the prediction is wrong, it updates the weights to reduce the error.If the prediction is correct, it does same as usual.



**Weights Updated Only When the Prediction is Incorrect**=> because learning is needed only when a mistake happens.


If prediction is correct: The model is already giving the right output.So, no change is needed.Updating weights may spoil the correct decision.


If prediction is incorrect:The model made an error.So, weights are updated to reduce this error next time.This helps the perceptron learn better decision boundaries.



# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:
A single-layer perceptron fails to solve the XOR problem because XOR data is not linearly separable.This means no single straight line can separate the output classes of XOR correctly.But a single-layer perceptron can create only a linear decision boundary.So, it cannot classify XOR correctly.

From XOR truth table we can see :

Class 0 → (0,0), (1,1)

Class 1 → (0,1), (1,0)


These points are placed diagonally, and no single straight line can separate them.So, XOR is not linearly separable ,So single-layer perceptron fails.


This limitation shows that => to solve non-linearly separable problems, we need multilayer neural networks.
Because: Multilayer networks have hidden layers.Hidden layers can create non-linear decision boundaries.This allows the model to learn complex patterns like XOR.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:

The Perceptron Learning Rule is a method used to update the weights of a perceptron so that it can learn from its mistakes.

W new= W old+ eta(y-y`)Xi

here,

W= weight

eta= learning rate

y= target output

y`= predicted output

The perceptron takes input and gives an output.It compares the predicted output with the correct output.If the prediction is wrong, it adjusts the weights.This adjustment reduces the error and improves future predictions.So, the perceptron learns by correcting its mistakes.



Why Can Adjusting One Weight Affect the Classification of Other Points:

This happens because all data points share the same weights.
When one weight is changed, the decision boundary also changes.This new line can change the classification of many data points, not just one.


In a 2D graph, the perceptron decision boundary is:

w1x1 + w2x2 + b = 0

Changing any weight (w₁/w₂) changes the slope and position of the line.

So, the dividing line moves or rotates.

As a result, some other points may move to the other side of the line, changing their class.

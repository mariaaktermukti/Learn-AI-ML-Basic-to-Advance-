# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write** Answer 01:

Gradient Descent is a optimization algorithm that is used in ML and DL to minimize the loss function by updating model parameters like weights and bias.

**Main Idea: ** let,i am somewhere on a hill wearing a blindfold. At the bottom of the valley is the optimal point or plane surface where i need to reach. Since i cannot see, the only info that i can use is the slope of the ground. The gradient tells me in what direction it is mostly downward, so i make a small step towards it. This whole process happens continuously, so after taking multiple steps in the direction that i believe goes down the most, i eventually stop when i feel that the ground is flat. In this analogy, the error in machine learning is the hill, and making a step is what gets the model closer to my end-goal.

**How It Helps Minimize Loss**:

 => We start with random initial parameters.

 => then calculate the gradient of the loss function at all parameters.

=> Then update the parameters by moving a small step in the opposite direction of the gradient. and repeat the whole process until the loss stops decreasing, meaning we have reached a minimum.


**Role of learing rate (eta):** it basically control how much the weights and bias are updated in each step. It also controls the step size of parameter updates.it also decides how fast or how slow the model can learns.It  affects how fastly the model reaches the minimum loss.

If Learning Rate is Too High: If the learning rate is too high, the model learns too fast, and it may miss the correct minimum point, causing unstable learning.

If Learning Rate is Too Low: If the learning rate is too low, the model learns too slowly, so it takes much more time to reach the optimal solution.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:
Softmax activation function is a non linear function that converts the output of a neural network into a probability distribution for multiple classes, where output range is (0,1), sum of outputs=1.

Why we useed in multi-class classification problems:It is used because it turns outputs into probabilities for all classes, making it easy to pick the most likely class in multi class problems.

 How is it different from sigmoid:

 Sigmoid is mainly used for binary classification, where there are only two classes, and it gives the probability for a single class between 0 and 1.
 On the other hand,softmax is used for multi class classificstion, where there are more than two classes and it gives probabilities for all classes at the same time and the sum of all probabilities is 1.

 Sigmoid predicts each class independently.
 On the other hand, softmax consider all classes together and chooses the class with the highest probability as the predicted class.
